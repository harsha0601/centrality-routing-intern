```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.cm as cm

G = nx.erdos_renyi_graph(20, 0.3, seed=42)
```

```python
deg = nx.degree_centrality(G)
bet = nx.betweenness_centrality(G)
clo = nx.closeness_centrality(G)

print("=== Degree Centrality ===")
for node, val in sorted(deg.items(), key=lambda x: -x[1]):
    print(f"Node {node}: {val:.4f}")

print("\n=== Betweenness Centrality ===")
for node, val in sorted(bet.items(), key=lambda x: -x[1]):
    print(f"Node {node}: {val:.4f}")

print("\n=== Closeness Centrality ===")
for node, val in sorted(clo.items(), key=lambda x: -x[1]):
    print(f"Node {node}: {val:.4f}")

eig = nx.eigenvector_centrality(G)

print("\n=== Eigenvector Centrality ===")
for node, val in sorted(eig.items(), key=lambda x: -x[1]):
    print(f"Node {node}: {val:.4f}")
```

```python
scores = [bet[n] for n in G.nodes()]
pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(10,7))

nodes = nx.draw_networkx_nodes(
    G,
    pos,
    node_color=scores,
    cmap=cm.Blues,
    node_size=700
)

nx.draw_networkx_edges(G, pos, alpha=0.5)
nx.draw_networkx_labels(G, pos)

plt.colorbar(nodes, label="Betweenness Centrality")

plt.title("Betweenness Centrality Graph")

plt.savefig("../results/week1_betweenness_graph.png")

plt.show()
```

## Observation

Nodes with higher betweenness centrality act as bridge nodes in the network. These nodes are important for communication flow and can improve routing efficiency in opportunistic networks.
